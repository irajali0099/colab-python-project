```python
def environmental_condition(temperature, humidity, rainfall):

    if temperature >= 20 and temperature <= 30 and humidity >= 60 and rainfall >= 100:
        return "Suitable environmental condition"
    elif temperature >= 15 and temperature <= 35:
        return "Moderately suitable environmental condition"
    else:
        return "Less suitable environmental condition"
```

```python
import pandas as pd

df = pd.read_csv("Crop_recommendation.csv")

print("Dataset loaded successfully!")
print("Number of records:", len(df))
print("Number of columns:", len(df.columns))
```

```python
features = [
    "N",
    "P",
    "K",
    "temperature",
    "humidity",
    "ph",
    "rainfall"
]

print("Selected Features:")
for feature in features:
    print(feature)
```

```python
def crop_decision(N, P, K, temperature, humidity, ph, rainfall):

    # Soil nutrient condition
    nutrient_average = (N + P + K) / 3

    if nutrient_average >= 60:
        soil_condition = "High"
    elif nutrient_average >= 30:
        soil_condition = "Medium"
    else:
        soil_condition = "Low"

    # Environmental condition
    if (
        temperature >= 20
        and temperature <= 30
        and humidity >= 60
        and rainfall >= 100
        and ph >= 5.5
        and ph <= 7.5
    ):
        environment_condition = "Suitable"

    elif (
        temperature >= 15
        and temperature <= 35
        and humidity >= 40
        and rainfall >= 60
    ):
        environment_condition = "Moderate"

    else:
        environment_condition = "Less Suitable"

    # Final agricultural decision
    if environment_condition == "Suitable" and soil_condition == "High":
        decision = "Highly Suitable Agricultural Condition"

    elif environment_condition == "Suitable":
        decision = "Suitable Agricultural Condition"

    elif environment_condition == "Moderate":
        decision = "Moderately Suitable Agricultural Condition"

    else:
        decision = "Less Suitable Agricultural Condition"

    return soil_condition, environment_condition, decision
```

```python
result = crop_decision(
    90,
    40,
    40,
    25,
    70,
    6.5,
    150
)

print("Soil Condition:", result[0])
print("Environmental Condition:", result[1])
print("Final Decision:", result[2])
```

```python
for i in range(10):

    row = df.iloc[i]

    soil_condition, environment_condition, decision = crop_decision(
        row["N"],
        row["P"],
        row["K"],
        row["temperature"],
        row["humidity"],
        row["ph"],
        row["rainfall"]
    )

    print("\nRecord:", i + 1)
    print("Actual Crop:", row["label"])
    print("Soil Condition:", soil_condition)
    print("Environmental Condition:", environment_condition)
    print("Decision:", decision)
```

```python
def recommend_crop(N, P, K, temperature, humidity, ph, rainfall):

    if N > P and N > K and temperature >= 20 and humidity >= 60:
        return "Nitrogen-rich warm condition"

    elif P > N and P > K and rainfall >= 100:
        return "Phosphorus-rich high-rainfall condition"

    elif K > N and K > P and temperature >= 20:
        return "Potassium-rich warm condition"

    elif ph >= 6 and ph <= 7.5 and rainfall >= 80:
        return "Balanced soil and suitable rainfall condition"

    else:
        return "General agricultural condition"
```

```python
for i in range(10):

    row = df.iloc[i]

    recommendation = recommend_crop(
        row["N"],
        row["P"],
        row["K"],
        row["temperature"],
        row["humidity"],
        row["ph"],
        row["rainfall"]
    )

    print("\nRecord:", i + 1)
    print("Actual Crop:", row["label"])
    print("Rule-Based Recommendation:", recommendation)
```

```python
results = []

for i in range(10):

    row = df.iloc[i]

    soil_condition, environment_condition, decision = crop_decision(
        row["N"],
        row["P"],
        row["K"],
        row["temperature"],
        row["humidity"],
        row["ph"],
        row["rainfall"]
    )

    results.append([
        i + 1,
        row["label"],
        soil_condition,
        environment_condition,
        decision
    ])

results_df = pd.DataFrame(
    results,
    columns=[
        "Record",
        "Actual Crop",
        "Soil Condition",
        "Environmental Condition",
        "Rule-Based Decision"
    ]
)

display(results_df)
```

```python
category_counts = results_df["Environmental Condition"].value_counts()

print(category_counts)
```

```python
def rule_based_crop(N, P, K, temperature, humidity, ph, rainfall):

    if (
        N > 80
        and temperature > 20
        and humidity > 60
        and rainfall > 150
    ):
        return "rice"

    elif (
        N > 70
        and temperature >= 20
        and temperature <= 30
        and rainfall >= 60
    ):
        return "maize"

    elif (
        K > 40
        and temperature >= 20
        and rainfall < 150
    ):
        return "cotton"

    elif (
        ph >= 6
        and ph <= 7
        and temperature >= 15
        and temperature <= 30
    ):
        return "chickpea"

    else:
        return "general"
```

```python
comparison = []

for i in range(10):

    row = df.iloc[i]

    predicted_crop = rule_based_crop(
        row["N"],
        row["P"],
        row["K"],
        row["temperature"],
        row["humidity"],
        row["ph"],
        row["rainfall"]
    )

    actual_crop = row["label"]

    if predicted_crop == actual_crop:
        result = "Match"
    else:
        result = "No Match"

    comparison.append([
        i + 1,
        actual_crop,
        predicted_crop,
        result
    ])

comparison_df = pd.DataFrame(
    comparison,
    columns=[
        "Record",
        "Actual Crop",
        "Predicted Crop",
        "Result"
    ]
)

display(comparison_df)
```

```python
matches = (comparison_df["Result"] == "Match").sum()
total_records = len(comparison_df)

print("Total Records:", total_records)
print("Matching Decisions:", matches)
print("Incorrect Decisions:", total_records - matches)
```

```python
matching_percentage = (matches / total_records) * 100

print("Matching Percentage:", round(matching_percentage, 2), "%")
```

```python
result_counts = comparison_df["Result"].value_counts()

import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))

result_counts.plot(kind="bar")

plt.title("Rule-Based Crop Decision Results")
plt.xlabel("Decision Result")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))

category_counts.plot(kind="bar")

plt.title("Agricultural Environmental Conditions")
plt.xlabel("Environmental Condition")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
